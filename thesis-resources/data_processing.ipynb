```python
txt_file_path = "plastics.txt"
csv_file_path = "plastics.csv"

# Read the file line by line
with open(txt_file_path, "r", encoding="utf-8") as file:
    lines = file.readlines()

# Initialize an empty list to store plastic-manufacturer pairs
plastic_data = []

# Process each line
for line in lines:
    line = line.strip()  # Remove extra whitespace and newlines

    parts = line.rsplit("(", 1)
    name = parts[0].strip()

    manufacturers = parts[1][:-1].strip()
    manufacturers_list = manufacturers.split('/')

    for manufacturer in manufacturers_list:
        manufacturer = manufacturer.strip()
        plastic_data.append((name, manufacturer))


# Convert to Pandas DataFrame
df = pd.DataFrame(plastic_data, columns=["Name", "Manufacturer"])

# Save to CSV
df.to_csv(csv_file_path, index=False, sep=",")

print(f"✅ Data successfully saved to {csv_file_path}")

# Display first few rows
df.head()
```

```python
# Define a function to categorize disc type based on speed
def categorize_disc(speed):
    if pd.isna(speed):
        return "Unknown"
    elif speed < 4:
        return "Putt & Approach"
    elif speed < 6:
        return "Midrange"
    elif speed < 10:
        return "Fairway Driver"
    else:
        return "Distance Driver"

# Apply the function to create the new 'Type' column
discs_df["Type"] = discs_df["Speed"].apply(categorize_disc)

# Display first few rows to verify
discs_df.head()

# Save the updated DataFrame
discs_df.to_csv("discs.csv", index=False, sep=",")

print("✅ discs.csv successfully updated with Type column!")
```

```python
# Load your data
discs_df = pd.read_csv("discs.csv")
plastics_df = pd.read_csv("plastics.csv")
manufacturers_df = pd.read_csv("manufacturers.csv")

# Create sets of unique manufacturers from each DataFrame
discs_manufacturers = set(discs_df["Manufacturer"].dropna().unique())
plastics_manufacturers = set(plastics_df["Manufacturer"].dropna().unique())
known_manufacturers = set(manufacturers_df["Name"].dropna().unique())

# Check discs_df manufacturers
print("Manufacturers in discs.csv not found in manufacturers.csv:")
for m in sorted(discs_manufacturers):
    if m not in known_manufacturers:
        print(" -", m)

# Check plastics_df manufacturers
print("\nManufacturers in plastics.csv not found in manufacturers.csv:")
for m in sorted(plastics_manufacturers):
    if m not in known_manufacturers:
        print(" -", m)
```

```python
# Replace incorrect names so they match with other datasets
plastics_df['Manufacturer'] = plastics_df['Manufacturer'].replace({
    "Latitude 64°": "Latitude 64",
    "Lone Star": "Lone Star Disc",
    "Legacy": "Legacy Discs"
})

# Save the updated DataFrame back to CSV
plastics_df.to_csv('plastics.csv', index=False)
```

```python
manufacturers_df = pd.read_csv("manufacturers.csv", encoding="utf-8-sig")
discs_df = pd.read_csv("discs.csv", encoding="utf-8-sig")
plastics_df = pd.read_csv("plastics.csv", encoding="utf-8-sig")

# Create a dictionary mapping manufacturer names to IDs
manufacturer_dict = dict(zip(manufacturers_df["Name"], manufacturers_df["ID"]))

# Print dictionary for verification
print("🔹 Manufacturer Dictionary:", manufacturer_dict)

# Replace manufacturer names with their corresponding ID in the discs dataset
discs_df["Manufacturer"] = discs_df["Manufacturer"].map(manufacturer_dict)

# Display updated dataset
discs_df.head()

# Replace manufacturer names with their corresponding ID in the plastics dataset
plastics_df["Manufacturer"] = plastics_df["Manufacturer"].map(manufacturer_dict)

# Display updated dataset
plastics_df.head()

# Save the updated CSV files
discs_df.to_csv("discs.csv", index=False, sep=",")
plastics_df.to_csv("plastics.csv", index=False, sep=",")

print("✅ discs.csv and plastics.csv successfully updated with manufacturer IDs!")
```
